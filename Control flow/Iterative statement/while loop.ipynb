```python
#Iterative: If we want to execute group of statements, multiple times OR n number of times we should go for iterative statement
#1.While #2.For
#step1: Intialization statement(always execute 1 time)   i=1
#step2: Conditional Statement                            i<=5
#step3: Icreament/Decreament                             i=i+1
```

```python
#Print 1 to 5 (1,2,3,4,5)
i = 1 
while i <= 5:
    print(i)
    i = i + 1
```

```python
#Print 1 to 20
i=1 
while i<=20:
    print(i)
    i=i+1
```
