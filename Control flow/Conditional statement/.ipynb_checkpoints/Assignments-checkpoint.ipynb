```python
#Check if a number is divisible by 7 or not
num=int(input("Enter a Number"))
if num%7==0:
    print("Number is divisible by 7")
else:
    print("Number is not divisible by 7")
```

```python
#Check if a number is positive or not
num=int(input("Enter a Number:"))
if num>0:
    print("Number is Positive")
elif num<0:
    print("Number is Negative")
else:
    print("Number is Zero")
```

```python
#Program to take single digit number from the keyboard and print in english
def number_to_word(num):
    if num==0:
        return "Zero"
    elif num==1:
        return "One"
    elif num==2:
        return "Two"
    elif num==3:
        return "Three"
    elif num==4:
        return "Four"
    elif num==5:
        return "Five"
    elif num==6:
        return "Six"
    elif num==7:
        return "Seven"
    elif num==8:
        return "Eight"
    elif num==9:
        return "Nine"
    else:
        return "Invalid Number"
#Get User input
num=int(input("Enter the sigle digit number:"))
#convert the number to its word equilant
word=number_to_word(num)
#print the result
print(word)
```
