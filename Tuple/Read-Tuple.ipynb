```python
# Read-Tuple:

# Read/Accessing elements using Indexing
# Can also use Slice operator to read elements 
```

```python
ids = (101,102,103,104,105,106)
#index  0   1   2   3   4   5
print(ids[0])
print(ids[1])
print(ids[2])
```

```python
ids = (101,102,103,104,105,106)
#index  0   1   2   3   4   5
print(ids[:])
print(ids[1:4])
print(ids[5:20])
```

```python
enames=("Rahul","sonia","priyanka","modhi")
#           0       1         2        3
#print 'priyanka' using indexing and slicing

print(enames[2])
print(enames[2:3])
```

```python
# Diff b/w List and Tuple
# List:
names=['rahul','sonia','priyanka']
names[1]='sonia gandhi'
print(names)
```

```python
# Tuple:
names=('rahul','sonia','priyanka')
names[1]='sonia gandhi'
print(names) # once object created, cannot change its content
```

```python
# Tuple methods
# index()
# count()
```

```python
names=('rahul','sonia','priyanka','sonia','sonia')
print(names.count('sonia'))
```

```python
names=('rahul','sonia','priyanka','sonia','sonia')
print(names.index('sonia'))
```

```python
names=['rahul','sonia','priyanka','sonia','sonia']
names.sort()
print(names) # List
```

```python
names=('rahul','sonia','priyanka','sonia','sonia')
print(tuple(sorted(names))) # Tuple
```

```python
numbers=(101,344,566,3,55,88,90,99)
result=sorted(numbers)
print(tuple(result))
```
