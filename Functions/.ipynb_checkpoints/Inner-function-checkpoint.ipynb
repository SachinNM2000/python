```python
#How to invoke inner function, outside the function?

def outer():
    
    def inner():
        print("Inner function")
     
    #return 100,200 //(100, 200)
    #return "Hello, Good Morning" //Hello, Good Morning
    return inner   #returns the function reference(inner=function name)

inner_fun_ref=outer()
inner_fun_ref()
inner_fun_ref()  
```
