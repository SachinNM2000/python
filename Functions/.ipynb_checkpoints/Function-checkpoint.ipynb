```python
#def: (keyword) -used to define a function
#function_name: name of the function which should be descriptive
#parameters: input values that the function can accept
#return: used to return value from the function
```

```python
# How to invoke the function?

# -by calling function name.
# example:
def add():
    print("Addition")  
            
add()            
add()             # can invoke/call the functions n number of times 
```

```python
def add(a,b):   # parameters(takes n number)
    print(a+b)

add(10,20)   # to recieve the arguments, parameters are required
add(100,200)
add("Hello,","GM")
```

```python
def add(a,b,c,d):
    result=a+b+c+d     # Just add the four numbers
    #executing some logic
    return result         
    #return some value
output=add(1,2,3,4)    # This will now return 10 (1 + 2 + 3 + 4)
print(output)
```

```python
#Function- need to verify the given number is 3 digit number or not.

def three_digit(num):
    if (num>=100 and num<=999):
        print("Given number is 3 gidit number")
    else:
        print("Given number is not a 3 gidit number")
        
three_digit(345)
```

```python
#Function- need to verify the given number is even od odd number.

def even_odd(num):
    if (num%2==0):
        print("Given number is even number")
    else:
        print("Given number is odd number")

even_odd(26)
```

```python
# Parameters and Arguments
#eg:1

def add():
    pass

add(1,2,3)   #while passing arguments it requires the parameters
```

```python
#eg:2

def add(a,b,c):  #parameters (positinal arguments)
    print(a+b+c)

add(1,2,3)       #arguments
```
