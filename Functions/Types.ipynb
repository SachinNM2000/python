```python
#Types of Functional arguments

#1. Positional arguments
#2. Keyword arguments
#3. Default arguments
#4. Variable length arguments
```

```python
#Positinal Arguments: Arguments that are passed in the same order as parameters 
#                     defined in the function.

def num(a,b):       
    
    print(a+b)
    print(a-b)
    
num(10,20)   # 10 is passed to a, 20 is passed to b (passed based on order)
```

```python
def num(a,b,c):       
    
    print(a+b,c)
    print(a-b,c)
    
num(10,20)
```

```python
#Keyword Arguments: Argumnets are passed using parameters names, and the order doesn't matter

def num(a,b):       
    
    print(a-b)
    
num(a=10,b=20)
num(b=20,a=10)  #passed using the parameter name
```

```python
#Default Arguments: can set default values for the parameters, 
#                   if no argument is provided the default value is used.

#eg.1:

def add(a,b,c=100):
    
    print(a+b+c)
    
add(10,20,30)
```

```python
#eg.2:

def add(a,b,c=100):
    print(a+b+c)
    
add(10,20)  #default value is used
```

```python
#Variable length Arguments: 
#     *args
#    **kwargs
```

```python
# *args: It collects all the positional arguments into tuple.
#        can access elements by index.

#syntax: 
#        def function_name(*args):
```

```python
def person(name, **data):
    print(name)
    print(data) 
    

person('sachin',age=25,cit='mysuru',ph=9606217683)
```

```python
def person(name, **data):
    print(name)
    
    for i,j in data.items():
        print(i,j)
    

person('sachin',age=25,city='mysuru',ph=9606217683)
```
