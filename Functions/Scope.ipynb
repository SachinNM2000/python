```python
#Local variable and Global variable
```

```python
def inside():
    x=5
    print("inside function", x)
inside()
print("outside", x)
```

```python
a=10               #Global variable

def something():
    global a
    a=15         # a value updated
    print("in function",a)
    
    
    
something()
print("outside",a)
```

```python
a=10               # Global variable
print(id(a))
def something():
    a=9            # Local variable 
    
    x=globals()['a']
    print(id(x))
    print("in fun",a)
    globals()['a']=15
    
     
something()

print("outside",a)
```

```python
a=100     #Global

def fun_one():
    print(a)
def fun_two():
    print(a)
    
fun_one()
fun_two()
```

```python
# how to convert local variable to global variable
a=10
def fun_one():
    global a
    a=100
    
def fun_two():
    print(a)
    
fun_one()
fun_two()
```
