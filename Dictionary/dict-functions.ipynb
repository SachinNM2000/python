```python
#divct-functions:

#get(key)     : returns the values associated with key
#pop(key)     : removes the specified entry from the dict 
#popitem() : removes the random entry from the dict(*key:values*=Entry)
#keys()    : returns all the keys presented in the dict
#values()  : returns all the values presented in the dict
#items()   : returns all key:value tuple pairs presented in the dict
```

```python
emp={'id':101,'name':'rahul','loc':'wayanadu'}

print(emp.get('id'))
print(emp.get('name'))
```

```python
emp={'id':101,'name':'rahul','loc':'wayanadu'}

emp.pop('loc')
print(emp)
```

```python
emp={'id':101,'name':'rahul','loc':'wayanadu','age':25}

emp.popitem()
print(emp)
```

```python
emp={'id':101,'name':'rahul','loc':'wayanadu'}

print(emp.keys())
print(emp.values())
print(emp.items())
```

```python
#can iterate using loops

emp={'id':101,'name':'rahul','loc':'wayanadu'}

for key in emp.keys():
    print(key)
    
for value in emp.values():
    print(value)
        
for item in emp.items():
    print(item)
```
