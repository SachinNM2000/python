```python
#dict : 
#       duplicate keys are not allowed but values are allowed
#       indexing and slicing is not applicable
#       insertion order is preserved
#       Mutable object
```

```python
#Create: 
user={}
emp={'id':101,'name':'rahul','loc':'wayanadu'}
d1=dict()

print(user)
print(emp)
print(d1)
```

```python
#Read: - how access dict values? ===== using keys 
                                                #(dict['key'])
```

```python
emp={'id':101,'name':'rahul','loc':'wayanadu'} 

print(emp['id'])
print(emp['name'])
print(emp['loc'])
```

```python
emp={'id':101,'name':'rahul','loc':'wayanadu'}
print(emp['email']) # KeyError(specified key is not present in the object)
```

```python
#Update: 
emp={'id':101,'name':'rahul','loc':'wayanadu'}

emp['name']='rahul gandhi'
print(emp)

emp['email']='rahul@gmail.com'
print(emp)
```

```python
#Delete: 

#(1)del: removes the specified key and its values
emp={'id':101,'name':'rahul','loc':'wayanadu'}

del emp['id']
print(emp)
```

```python
#(2)clear: removes the all properties(key:values) the set.
emp={'id':101,'name':'rahul','loc':'wayanadu'}

emp.clear()
print(emp)
```

```python
#example: 
emp={'id':101,'name':'rahul','loc':'wayanadu'}

del emp['id']
print(emp)

emp.clear()
print(emp)

del emp      # deleting complete dict object
print(emp)
```
