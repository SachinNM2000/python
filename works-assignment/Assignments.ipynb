```python
# Function to find the 3rd largest number
def find_num(numbers):
    numbers.sort(reverse=True)  # Sort the list in descending order
    return numbers[2]  # Return the number at index 2 (the 3rd largest number)

# Correct list of numbers
numbers = [10, 20, 30, 40, 50]

# Call the function and print the result
result = find_num(numbers)
print(result)
```

```python
# Function to find the largest numbernumbers.
def large_num(numbers):
    return max(numbers)
numbers=[10,20,30]
result=large_num(numbers)
print(result)
```

```python
# Function to find the minimum numbernumbers.
def large_num(numbers):
    return min(numbers)
numbers=[10,20,30]
result=large_num(numbers)
print(result)
```

```python
#single level inheritance
class A:
    def fun1(self):
        print("fun1 is working")
        
    def fun2(self):
        print("fun2 is working")
        
class B(A):
    def fun3(self):
        print("fun3 is working")
        
    def fun4(self):
        print("fun4 is working")
        

b1=B()
b1.fun1()
b1.fun2()
b1.fun3()
b1.fun4()
```

```python
#multi-level inheritance
class A:
    def fun1(self):
        print("fun1 is working")
        
    def fun2(self):
        print("fun2 is working")
        
class B(A):
    def fun3(self):
        print("fun3 is working")
        
    def fun4(self):
        print("fun4 is working")

class C(B):
    def fun5(self):
        print("fun5 is working")
       
b1 = B()
c1 = C()

c1.fun1()
c1.fun2()
c1.fun3()
c1.fun4()
c1.fun5()
```

```python
#multiple inheritance
class A:
    def fun1(self):
        print("fun1 is working")
        
    def fun2(self):
        print("fun2 is working")
        
class B():
    def fun3(self):
        print("fun3 is working")
        
    def fun4(self):
        print("fun4 is working")

class C(A,B):
    def fun5(self):
        print("fun5 is working")

        
a1 = A()        
b1 = B()
c1 = C()

c1.fun1()
c1.fun2()
c1.fun3()
c1.fun4()
c1.fun5()
```
