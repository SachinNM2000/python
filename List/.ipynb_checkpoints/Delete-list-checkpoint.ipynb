```python
# Delete-list

# pop: it removes the last element from the list and returns the removed value.

enames=['rahul','sonia','priyanka','modhi']
enames.pop()
print(enames)
```

```python
enames=['rahul','sonia','priyanka','modhi']
values=enames.pop()
print(enames)
print(values)
```

```python
enames=['modhi','rahul','sonia','priyanka',]
#index     0       1        2         3
enames.pop(0)
print(enames)
```

```python
# remove(): removes the specified element  from the list.

enames=['rahul','sonia','priyanka','modhi']
enames.remove(3) # can give str 'amith'
print(enames)
```

```python
enames=['rahul','sonia','priyanka','modhi']
enames.remove('modhi')
print(enames)
```

```python
# clear(): clears all the values from the list.
ids=[101,102,103,104,105]
ids.clear() # clear takes no arguments
print(ids)
```
