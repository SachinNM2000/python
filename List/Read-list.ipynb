```python
#How to read list elements
ids=[101,102,103,104]
#     0  1   2   3
enames=['Rahul','Sonia','priyanka']
#         0        1        2

#ids index
print(ids[0])
print(ids[1])
print(ids[2])
print(ids[3])

#enames index
print(enames[0])
print(enames[1])
print(enames[2])
print(enames[5])  #index error
#accessing list elements using indexing
```

```python
enames=['Rahul','Sonia','Priyanka','Modhi','Amith']
#          0       1        2          3      4
#Iterating list objects using for and while loop

for ename in enames:
    print(ename)

i=0
while i<=4:
    print(enames[i])
    i=i+1
```
