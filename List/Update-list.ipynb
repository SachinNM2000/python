```python
#Update-list
# append(): is a list function/method to add element to end of the list.
ids=[101,102,103,104]
# add element to list
ids.append(105)
print(ids)
```

```python
# insert(): to add element to the specified index
ids=[101,102,103,104]
#     0   1   2   3
#syntax: specified (index value, element)
ids.insert(3,108)
print(ids)
```

```python
# extend(): to add a another list. Expecting sequence in the list (either LIST or TUPLE etc).
enames=['rahul','sonia','priyanka']
new_enames=[101]
enames.extend(new_enames)
print(enames)
```

```python
enames=['rahul','sonia','priyanka']
enames.extend('arvind')
print(enames)
```

```python
# sort(): If we want to apply sorting mechanism we need homogeneous elements.
l1=[101,102,'rahul','sonia']
print(l1.sort())
```

```python
# Default: Natural sorting order
# Number: ascending order 0 to 9
# Alphabet: a to z
```

```python
enames=['rahul','sonia','priyanka','akash','virat']
enames.sort()
print(enames)
```

```python
ids=[102,101,106,110,115,133]
ids.sort()
print(ids)
```

```python
# reverse(): reversing the order of elements in the list.

enames=['rahul','sonia','priyanka','akash','virat']
enames.reverse()
print(enames)
```

```python
# Natural sorting order - ascending 
enames=['rahul','sonia','priyanka','virat','anil']

# descending order 
enames.sort(reverse=True)
print(enames)
```
