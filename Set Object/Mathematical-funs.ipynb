```python
#Union() : returns all the elements 
a={10,20,30,40}
b={30,40,50,60}

print(a.union(b))

print(a|b)

result=a|b
print(result)
```

```python
#Intersection() : returns common elements
a={10,20,30,40}
b={30,40,50,60}

print(a.intersection(b))
print(a&b)
```

```python
#Difference : returns the set A elements excludes common elements
a={10,20,30,40}
b={30,40,50,60}

print(a.difference(b))
print(a-b)
```

```python
#Symmetric_differenece() : returns all the elements excludes common elements
a={10,20,30,40}
b={30,40,50,60}
 
print(a.symmetric_difference(b))
print(a^b)
```
