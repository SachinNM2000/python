```python
# Set Functions
#Create: set() copy()
#Read: 
#Update: add() update()
#         Mathematical functions: union() intersection() difference() symmetric_difference()
#Delete: remove() discard() clear() pop() 
```

```python
#set()
s1 = set()
print(s1)
print(type(s1))
```

```python
#add() - add one elementat a time to a set object
ids={101,102,103,104}
ids.add(105)
print(ids)
```

```python
#len() - is a inbuilt function in python
ids={101,102,103,104,105}
print(len(ids)) 
```

```python
#update() : Adds multiple elements to the set
#        Takes an iterable objectS(e.g., a list, tuple, or another set)  
#             and adds each element from that iterable to the set.
#        string is allowed in update(),- str is an iterable object in python so, 
#             each character is added individually to the set
ids={101,102,103,104,105}
ids.update({10,20,30})
print(ids)
```

```python
#example-1:
ids={101,102,103,'rahul'}
ids.update('sachin')
print(ids)  # set is an unordered
```

```python
enames={'rahul','sonia','priyanka'}
enames.update('sachin')
print(enames)
```

```python
#pop() : removes and return random element in the set
nums={10,20,30,40,50}
nums.pop()
print(nums)
```

```python
#remove : removes the specified element
nums={10,20,30,40,50}
nums.remove(30)
print(nums)
```

```python
# if specified element is not present in the set it returns the key error
nums={10,20,30,40,50}
nums.remove(60)
print(nums)
```

```python
#discard : removes the specified element 
nums={10,20,30,40,50}
nums.remove(30)
print(nums)
```

```python
#But, in discard if specified element is not present in the set it returns nothing
nums={10,20,30,40,50}
nums.remove(30)
print(nums)
```

```python
#Clear : it clears all the element in the set and returns empty set
nums={10,20,30,40,50}
nums.clear()
print(nums)
```
